# Data Consolidation Pipeline

### The code looks into your folder and finds all your CSV files and transforms individual data to clean unified database

```python
import pandas as pd
from pathlib import Path

path = Path("11-15_Patients")
# Combines your folder name and the *.csv pattern into a list of CSV files.
all_files = sorted(path.glob("*.csv"))

#The loop (processing each file)
## li-- Initializes an empty list to hold the data from each file temporarily.
li =[]

# This code transforms a messy folder of individual reports into a clean, unified database.
for filename in all_files:
    df = pd.read_csv(filename, sep=";", index_col = None, header=0)
    #Tagging the Data 
    df['patient_id'] = filename.stem
    li.append(df)

#Merging everything

combined_activity = pd.concat(li, axis= 0,ignore_index = True)
combined_activity.to_csv("Combined_Datasets_11-15.csv", index=False)
```

# Rounding the columns to 3 decimals


```python
cols=['calories','glucose','heart_rate','steps','basal_rate']
df[cols]=df[cols].round(3)
df.head()
```

# Checking for Outliers

```python
numeric_cols = [
    'glucose',
    'calories',
    'heart_rate',
    'steps',
    'basal_rate',
    'bolus_volume_delivered',
    'carb_input'
]
 # Q1 is value where 25% of data falls Q3 is value where 75% of data falls. IQR gived the 50% data range ,this is the most reliable part of your dataset because it ignores the extreme highs and extreme lows.
Q1 = df[numeric_cols].quantile(0.25)
Q3 = df[numeric_cols].quantile(0.75)
IQR = Q3 - Q1

outliers = (
    df[numeric_cols] < (Q1 - 1.5 * IQR)
) | (
    df[numeric_cols] > (Q3 + 1.5 * IQR)
)

outliers.sum()
```

# Replacing Outliers

## Using the 1.5×IQR rule to detect outliers and cap them to safe limits, ensuring your numeric clinical data stays clean, stable, and analysis‑ready without losing any rows.

```python
# 1. Define your fences upper and lower bounds 
lower_fence = Q1 - 1.5 * IQR
upper_fence = Q3 + 1.5 * IQR

# 2. Apply the cap (clipping i.e is easier way of handling if/else logic for many rows at once)
df_cleaned = df.copy()
# 3. Replacing lower outliers with lower fence and upper detected outlier with upper fence value
df_cleaned[numeric_cols] = df_cleaned[numeric_cols].clip(lower=lower_fence, upper=upper_fence, axis=1)

print("Outliers have been capped to the 1.5*IQR fences.")
```

## Standardizing Time zones If some data was recorded in UTC and some in EST, it strips those labels so they can be compared easily as "naive" local time and sort time zones in chronlogical order removing duplicates

```python
# Convert to datetime and strip timezone info
df['time'] = pd.to_datetime(df['time']).dt.tz_localize(None)

# Sort and remove duplicates (keep the first occurrence)
df = df.sort_values(['patient_id', 'time'])
df = df.drop_duplicates(subset=['patient_id', 'time'], keep='first').reset_index(drop=True)
```

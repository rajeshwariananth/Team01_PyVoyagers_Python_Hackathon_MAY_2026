```python
import pandas as pd
import numpy as np
import os

folder_path = "C:/Users/deept/OneDrive/Desktop/Python data"

files = os.listdir(folder_path)

cleaned_data = []

for file in files:
    if file.endswith(".csv"):
        
        file_path = os.path.join(folder_path, file)
        df = pd.read_csv(file_path)
        
        print(f"Processing: {file}")
        
        # 1. Rename columns
        df.columns = (
            df.columns
            .str.strip()
            .str.lower()
            .str.replace(" ", "_")
            .str.replace("-", "_")
            .str.replace("/", "_")
        )
        
        # 2. Change data types
        for col in df.columns:
            df[col] = pd.to_numeric(df[col], errors="ignore")
        
        # 3. Round numeric columns to 3 decimals
        numeric_cols = df.select_dtypes(include=["int64", "float64"]).columns
        df[numeric_cols] = df[numeric_cols].round(3)
        
        # 4. Find outliers using IQR method
        outlier_cols = []

        for col in numeric_cols:
            Q1 = df[col].quantile(0.25)
            Q3 = df[col].quantile(0.75)
            IQR = Q3 - Q1
            
            lower_limit = Q1 - 1.5 * IQR
            upper_limit = Q3 + 1.5 * IQR
            
            df[col + "_outlier"] = (
                (df[col] < lower_limit) | 
                (df[col] > upper_limit)
            )
            
            if df[col + "_outlier"].sum() > 0:
                outlier_cols.append(col)
        
        # Add source file name
        df["source_file"] = file
        df["columns_with_outliers"] = ", ".join(outlier_cols)
        
        cleaned_data.append(df)

final_df = pd.concat(cleaned_data, ignore_index=True)

final_df.to_csv(
    "C:/Users/deept/OneDrive/Desktop/Python data/final_cleaned_all.csv",
    index=False
)

final_df.head()
```

```python
import pandas as pd
import numpy as np
import os

# Folder where your CSV files are stored
folder_path = "C:/Users/deept/OneDrive/Desktop/Python data"

# Get all files from folder
files = os.listdir(folder_path)

cleaned_data = []

for file in files:
    if file.endswith(".csv") and file != "final_cleaned_all.csv":
        
        file_path = os.path.join(folder_path, file)
        
        # Read semicolon-separated CSV file
        df = pd.read_csv(file_path, sep=";")
        
        print(f"Processing: {file}")
        
        # 1. Rename columns professionally
        df.columns = (
            df.columns
            .str.strip()
            .str.lower()
            .str.replace(" ", "_")
            .str.replace("-", "_")
            .str.replace("/", "_")
            .str.replace("_", " ")
            .str.title()
        )
        
        # 2. Change data types
        for col in df.columns:
            converted = pd.to_numeric(df[col], errors="coerce")
            if converted.notna().sum() > 0:
                df[col] = converted
        
        # Convert Time column to datetime if it exists
        if "Time" in df.columns:
            df["Time"] = pd.to_datetime(df["Time"], errors="coerce")
        
        # 3. Round numeric columns to 3 decimals
        numeric_cols = df.select_dtypes(include=["int64", "float64"]).columns
        df[numeric_cols] = df[numeric_cols].round(3)
        
        # 4. Find outliers using IQR method
        outlier_cols = []

        for col in numeric_cols:
            Q1 = df[col].quantile(0.25)
            Q3 = df[col].quantile(0.75)
            IQR = Q3 - Q1
            
            lower_limit = Q1 - 1.5 * IQR
            upper_limit = Q3 + 1.5 * IQR
            
            outlier_col_name = col + " Outlier"
            
            df[outlier_col_name] = (
                (df[col] < lower_limit) |
                (df[col] > upper_limit)
            )
            
            if df[outlier_col_name].sum() > 0:
                outlier_cols.append(col)
        
        # 5. Add source file name
        df["Source File"] = file
        df["Columns With Outliers"] = ", ".join(outlier_cols)
        
        cleaned_data.append(df)

# Combine all cleaned files
final_df = pd.concat(cleaned_data, ignore_index=True)

# Save final cleaned dataset
output_path = "C:/Users/deept/OneDrive/Desktop/Python data/final_cleaned_all.csv"

final_df.to_csv(output_path, index=False)

# Show preview
final_df.head()
```
